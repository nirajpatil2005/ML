Barplot

```python
#getting each row output so
px.bar(titanic,x=titanic['Pclass'],y=titanic['Age'],text_auto=True)
```

```python
import plotly.express as px
import pandas as pd

# Group the data by 'Pclass' and calculate the mean 'Fare'
aggregated_data = titanic.groupby('Pclass', as_index=False)['Age'].mean()

# Now plot the aggregated data using px.bar()
fig = px.bar(aggregated_data, 
             x='Pclass', 
             y='Age', 
             text_auto=True, 
             title='Average Fare by Passenger Class')

# Show the plot
fig.show()
```

```python
#we have to aggregate bar
aggregate_Age=titanic.groupby('Pclass',as_index=False)['Age'].mean()
aggregate_Age
```

```python
px.bar(aggregate_Age,x='Pclass',y='Age',text_auto=True)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plotting the KDE for Survived=0 and Survived=1 on the same plot
sns.kdeplot(titanic[titanic['Survived'] == 0]['Age'], label='Not Survived')
sns.kdeplot(titanic[titanic['Survived'] == 1]['Age'], label='Survived')

# Adding labels and title for better understanding
plt.title('Age Distribution by Survival Status')
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend()

# Show plot
plt.show()
```
