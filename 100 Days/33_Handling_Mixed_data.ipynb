```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('D:\\ML\\100 Days\\titanic_handling_mixed_data.csv')
df.shape
```

```python
df['number'].unique()
```

```python
fig=df['number'].value_counts().plot.bar()
fig.set_title('passengers Travelling with')
```

```python
#extract numerical part
df['number_numerical']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')
#extracr categorical part
df['number_cat']=np.where(df['number_numerical'].isnull(),df['number'],np.nan)
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
df['Cabin_num']=df['Cabin'].str.extract(r'(\d+)')#capture num part

df['Cabin_cat']=df['Cabin'].str[0]#captures the first letter
```

```python
df.head()
```

```python
df['Cabin_cat'].value_counts()
```

```python
#extract the last bit of ticket as number
df['ticket_num']=df['Ticket'].apply(lambda s:s.split()[-1])
df['ticket_num']=pd.to_numeric(
    df['ticket_num'],errors='coerce',downcast='integer'
)

#extract the first part of ticket as category
df['ticket_cat']=df['Ticket'].apply(lambda s:s.split()[0])
df['ticket_cat']=np.where(df['Ticket'].str.isdigit(),np.nan,df['ticket_cat'])
df.head(20)
```
