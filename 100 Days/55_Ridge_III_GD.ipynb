```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
x,y=load_diabetes(return_X_y=True)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=4)
```

```python
from sklearn.linear_model import SGDRegressor
```

```python
reg=SGDRegressor(penalty='l2',max_iter=500,eta0=0.1,learning_rate='constant',alpha=0.001)
```

```python
reg.fit(x_train,y_train)

y_pred=reg.predict(x_test)
print("R2:",r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```

```python
from sklearn.linear_model import Ridge
rridge=Ridge(alpha=0.001,max_iter=500,solver='sparse_cg')
```

```python
rridge.fit(x_train,y_train)
y_pred1=rridge.predict(x_test)
print("R2:",r2_score(y_test,y_pred1))
print(rridge.coef_)
print(rridge.intercept_)
```

```python
class MerazRidgeGD:
    
    def __init__(self,epochs,learning_rate,alpha):
        self.learning_rate=learning_rate
        self.epochs=epochs
        self.alpha=alpha
        self.coef_=None
        self.intercept_=None
        
    def fit(self,x_train,y_train):
        self.coef_=np.ones(x_train.shape[1])
        self.intercept_=0
        theta=np.insert(self.coef_,0,self.intercept_)#w
        x_train=np.insert(x_train,0,1,axis=1)
        for i in range(self.epochs):
            theta_der=np.dot(x_train.T,x_train).dot(theta)- np.dot(x_train.T,y_train) +self.alpha*theta
            theta=theta-self.learning_rate*theta_der
            
        self.coef_=theta[1:]
        self.intercept_=theta[0]
        print(self.intercept_,self.coef_)
        
    def predict(self,x_test):
        return np.dot(x_test,self.coef_)+self.intercept_
```

```python
reg=MerazRidgeGD(epochs=500,alpha=0.01,learning_rate=0.005)
```

```python
reg.fit(x_train,y_train)
```

```python
y_pred1=reg.predict(x_test)
print("R2:",r2_score(y_test,y_pred1))
print(reg.coef_)
print(reg.intercept_)
```
