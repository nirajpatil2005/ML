```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('student_clustering.csv')
print("Thes shape of the data is :",df.shape)
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['cgpa'],df['iq'])
```

```python
from sklearn.cluster import KMeans
```

```python
wcss=[]
for i in range(1,11):
    km=KMeans(n_clusters=i)
    km.fit_predict(df)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
x=df.iloc[:,:].values
km=KMeans(n_clusters=4)
y_means=km.fit_predict(df)
y_means
```

```python
x[y_means==0]
```

```python
x[y_means,1]
```

```python
plt.scatter(x[y_means==0,0],x[y_means==0,1],color='b')
plt.scatter(x[y_means==1,0],x[y_means==1,1],color='r')
plt.scatter(x[y_means==2,0],x[y_means==2,1],color='y')
plt.scatter(x[y_means==3,0],x[y_means==3,1],color='g')
```

        Kmeans On 3_D Data

```python
from sklearn.datasets import make_blobs

centroids=[(-5,-5,5),(5,5,-5),(3.5,-2.5,4),(-2.5,-2.5,-4)]
cluster_std=[1,1,1,1]
x,y=make_blobs(n_samples=100,cluster_std=cluster_std,centers=centroids,n_features=3,random_state=1)
```

```python
x
```

```python
import plotly.express as px
fig=px.scatter_3d(x=x[:,0],y=x[:,1],z=x[:,2])
fig.show()
```

```python
wcss=[]
for i in range(1,11):
    km=KMeans(n_clusters=i)
    km.fit_predict(x)
    wcss.append(km.inertia_)
```

```python
plt.plot(range(1,11),wcss)
```

```python
km=KMeans(n_clusters=4)
y_pred=km.fit_predict(x)
```

```python
df=pd.DataFrame()
df['col1']=x[:,0]
df['col2']=x[:,1]
df['col3']=x[:,2]
df['label']=y_pred
```

```python
fig=px.scatter_3d(df,x='col1',y='col2',z='col3',color='label')
fig.show()
```
