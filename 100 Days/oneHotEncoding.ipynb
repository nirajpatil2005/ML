```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv("covid_toy.csv")
df.head()
```

```python
df.sample(5)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop(columns='has_covid'),df['has_covid'],
                                               test_size=0.2)
```

```python
x_train
```

Aam Zindagi

```python
#adding simple impute to fever col
si=SimpleImputer()
x_train_fever=si.fit_transform(x_train[['fever']])

x_test_fever=si.transform(x_test[['fever']])
x_train_fever
```

```python
x_train_fever.shape
```

```python
#adding ordinal encoding-->cough
oe=OrdinalEncoder(categories=[['Mild','Strong']])
x_train_cough=oe.fit_transform(x_train[['cough']])
x_test_cough=oe.transform(x_test[['cough']])
x_train_cough.shape
```

```python
#one Hot encoding
ohe=OneHotEncoder(drop='first',sparse_output=False)
x_train_gender_city=ohe.fit_transform(x_train[['gender','city']])
x_test_gender_city=ohe.transform(x_test[['gender','city']])
x_train_gender_city
```

```python
x_train_gender_city.shape
```

```python
x_train_age=x_train.drop(columns=['gender','fever','cough','city']).values
x_test_age=x_test.drop(columns=['gender','fever','cough','city']).values
x_train_age
```

```python
x_train_age.shape
```

```python
x_train_transform=np.concatenate((x_train_age,x_train_fever,x_train_gender_city
                            ,x_train_cough),axis=1)
x_test_transform=np.concatenate((x_test_age,x_test_fever,x_test_gender_city
                            ,x_test_cough),axis=1)
x_train_transform
```

```python
x_train_transform.shape
```

Mentos Zindagi

```python
from sklearn.compose import ColumnTransformer
```

```python
df.head()
```

```python
trasformer=ColumnTransformer(transformers=[
                                            ('tnf1',SimpleImputer(),['fever']),
                                            ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
                                            ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city']),
                                            ],remainder='passthrough')
```

```python
trasformer.fit_transform(x_train)
```

```python
trasformer.fit_transform(x_train).shape
```
