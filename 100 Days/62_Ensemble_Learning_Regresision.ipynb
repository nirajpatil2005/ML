```python
from sklearn.datasets import load_diabetes
import numpy as np
```

```python
x,y=load_diabetes(return_X_y=True)
```

```python
x.shape,y.shape
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr=LinearRegression()
dt=DecisionTreeRegressor()
svr=SVR()
```

```python
estimators=[('lr',lr),('dt',dt),('svr',svr)]
```

```python
for estimator_name,estimator in estimators:
    score=cross_val_score(estimator,x,y,cv=10,scoring='r2')
    print(f"{estimator_name}: {np.round(np.mean(score),2)}")
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
vr=VotingRegressor(estimators=estimators)
scores=cross_val_score(vr,x,y,scoring='r2',cv=10)
print(f'voting regresor : {np.round(np.mean(scores),2)}')
```

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vr=VotingRegressor(estimators=estimators)
            scores=cross_val_score(vr,x,y,scoring='r2',cv=10)
            print(f'for i={i},j={j},k={k},{np.round(np.mean(scores),2)}')
```

                Using the same Algorithm

```python
dt1=DecisionTreeRegressor(max_depth=1)
dt2=DecisionTreeRegressor(max_depth=3)
dt3=DecisionTreeRegressor(max_depth=5)
dt4=DecisionTreeRegressor(max_depth=7)
dt5=DecisionTreeRegressor(max_depth=None)
```

```python
estimators=[('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
```

```python
for estimator_name,estimator in estimators:
    scores=cross_val_score(estimator,x,y,scoring='r2',cv=10)
    print(f"{estimator_name}:{np.round(np.mean(scores),2)}")
```

```python
vc2=VotingRegressor(estimators=estimators)
score=cross_val_score(vc2,x,y,cv=10,scoring='r2')
print(f"vc2:= {np.round(np.mean(score),2)}")
```

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vc2=VotingRegressor(estimators=estimators)
            score=cross_val_score(vc2,x,y,cv=10,scoring='r2')
            print(f"vc2:= {np.round(np.mean(score),2)}")
```
